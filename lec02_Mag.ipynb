### Структурное программирование. Циклы

В прошлом веке программы были еще довольно маленькие, и в них легко было разобраться. Но когда появились первые ЯП относительно высокого уровня, люди придумали команду go to, которая отправляла компьютер выполнять любую строчку, какую указывал программист, и внутренняя структура программ с этой командой становилась похожа на мешанину.

Поэтому ученый программист Эдсгер Дейкстра (https://en.wikipedia.org/wiki/Edsger_W._Dijkstra) написал целую статью про то, что давайте избавимся от этого go to и будем структурировать свой код, чтобы в нем не запутаться.

Так появилась парадигма структурного программирования. В рамках этой парадигмы предполагается, что мы, когда программируем, сперва разрабатываем строгий алгоритм, а потом пишем код в соответствии с этим алгоритмом, и никаких внезапных переходов на N-ю строчку.

В структурном программировании есть три основных понятия:
- последовательность
- ветвление
- цикл

Последовательное выполнение кода означает, что компьютер выполняет команду за командой в таком порядке, в котором они идут.

Ветвления - это условные инструкции, позволяющие выполнять один код при выполнении одних условий, и другой - при других. Мы с ними уже знакомы.

Циклы - то, что мы рассмотрим сейчас. Цикл - это блок инструкций в программе, которые повторяются или заданное количество раз, пока выполняется условие.

Таким образом, у нас есть две разновидности циклов:

1. **Определенный цикл** - когда мы заранее знаем, сколько раз нам нужно повториться, или перебираем конечный итерируемый объект. Например, когда нам нужно вывести таблицу умножения для всех цифр от 1 до 9.

2. **Неопределенный цикл** - когда мы должны повторять набор действий, пока выполняется какое-то условие. Например, мы ждем от пользователя, когда он введет цифру, а он вводит буквы, глупый человек. Мы никаким образом не можем предсказать, когда до него наконец дойдет, что нужно ввести цифру, поэтому цикл с заданным количеством итераций не подходит.

Определенный цикл - это цикл for, неопределенный - цикл while.

####Цикл for

В самом частом случае код такого цикла выглядит так:

      for i in range(n):
          command0
          command1
          ...

*i* - это переменная цикла, которая меняет свое значение, пока работает цикл. Вообще имя переменной может быть каким угодно, *i* используется просто по традиции. Именно из-за этой традиции объекты, которые можно перебрать поэлементно, называются **итерируемыми**, а единичный проход по циклу, соответствующий одному значению переменной цикла, называется **итерацией**.

range(n) - это результат применения функции range() к некоторому целому положительному числу *n*, который представляет собой отдельный вид итерируемого объекта, называемый **диапазоном**. На самом деле на месте range(n) в конструкции с for может стоять любой итерируемый объект - строка, список, кортеж, множество, словарь, итератор или генератор. Когда мы итерировали по строкам, спискам и множествам, мы по сути имели дело с частными случаями цикла for. Т.е. блок кода, идущий в цикле for после двоеточия (он называется **телом цикла**), выполняется столько раз, сколько значений принимает переменная цикла.

Итак, что же это за объект - диапазон? Это ряд чисел, который генерируется функцией range(). Если она используется с одним параметром *n*, то ряд начинается нулем и включает в себя последовательно все целые числа вплоть до *n*, но не включая само число *n*. Таким образом, ряд содержит ровно *n* элементов - это удобно! Именно потому, что нам удобно создавать с помощью параметра *n* ряд c *n* элементами, а программисты, как мы помним, начинают отсчет с нуля, а не с единицы, логично, что число *n* мы не должны включать в диапазон. Если же нам надо задать диапазон, который будет включать само *n*, мы должны, указывая параметр для range(), увеличить *n* на единицу, то есть написать:

        for i in range(n+1):
            ...

Полный синтаксис функции range() включает три параметра, аналогичных тем параметрам, которые мы использовали при создании срезов:

    range(start, stop, step)

Только начало, конец и шаг в случае диапазона передаются через запятую, а не через двоеточие, потому что здесь это аргументы функции. Все аргументы функции передаются через запятую, помним это.

Как и в случае срезов, мы можем задать отрицательный шаг, чтобы создать нисходящий ряд чисел. В этом случае так же конец диапазона (т.е. меньший параметр) не будет входить в полученный диапазон, тогда как начало (больший параметр) будет входить в него. Будьте внимательны, не запутайтесь!

Кроме того, мы можем задать range() только c двумя параметрами, и тогда они будут интерпретироваться как начало и конец диапазона, соответственно:

    range(start, stop)

Ниже дан пример конструкции for i in range() с тремя параметрами range().

```python
for i in range(1, 10, 2):
    print(i)
```

####Цикл while

Полный синтаксис цикла while выглядит следующим образом:

      while <condition>:
          command
          command
      else:
          command

Условие может быть каким угодно выражением, имеющим логическое значение или переводимым в него. Можно даже непосредственно указывать True, но тогда, чтобы закончить цикл, нужно применять альтернативный способ выхода (см. ниже), потому что True - это условие, которое всегда верно.

Цикл while очень похож на if с той лишь разницей, что в if инструкции при выполнении условия срабатывают один раз, а в цикле while они проходятся снова и снова, пока условие не будет нарушаться. Однако инструкция else все равно срабатывает только один раз в конце работы цикла, т.е. впервые, когда перестало выполняться условие.

На самом деле в цикле for тоже можно задать конечную инструкцию else, которая будет срабатывать один раз после выполнения всех необходимых итераций тела цикла. Однако else используется с циклом for крайне редко.

Ниже дан пример цикла while с инструкцией else.

```python
number = input()
while not number.isdigit():
    number = input('Введите число!\n')
else:
    print('Наконец-то число!')
```

Если мы передаем в качестве условия на выполнение цикла while объект другого типа, то происходит неявное преобразование типов, и этот объект превращается в bool. Как мы уже говорили, в случае чисел любое число, кроме нуля, превращается в True, а ноль - в False; в случае строк - любая строка, кроме пустой, превращается в True, а пустая - в False; в случае других итерируемых объектов это работает так же, как со строками.

Таким образом, абсолютно корректным (и демонстрирующим хороший стиль) является следующий код:

```python
x = int(input())
summa = 0
while x:
    summa += x
    x = int(input())
print(summa)
```

####Что еще нужно знать о циклах?

Наряду с условными инструкциями циклы - основные способы управления работой программы.

Циклы можно вкладывать друг в друга (как и условные инструкции). Обратите внимание, что индентинг (отступы) выравнивает команды и помогает визуально отличить, где какая часть программы. Иногда бывает несложно спутать, к чему относится else: смотрите, на одном уровне с какой инструкцией он находится. В IDE обычно проводятся вертикальные линии, чтобы удобно было это отслеживать.

    while <condition1>:
        if <condition2>:
            for i in <iterable>:
                command0  # эта команда - в теле цикла for!
        else: # этот else относится к if!
            command1
            command2
        command3 # эта команда - в теле цикла while!
    else: # этот else относится к while!
        command4


Будьте внимательны! При использовании вложенных циклов for называйте переменную цикла по-разному, иначе программа будет работать некорректно. Такая ситуация называется **коллизией переменных**. Вот пример.

```python
# ХОРОШИЙ КОД: используем разные имена
total = 0
for outer in range(1, 3):
    for inner in range(1, 3):
        total += inner
    total *= outer
    print(total) #ожидаем на первой итерации (0 + 1 + 2) * 1 = 3, на второй (3 + 1 + 2) * 2 = 12
print(total == 12) #True - работает корректно

# ПЛОХОЙ КОД: коллизия переменных
total = 0
for i in range(3):
    for i in range(3):
        total += i
    total *= i
    print(total) #из-за того, что значение i меняется во внутреннем цикле, имеем другой результат
print(total == 12) #False - работает некорректно
```

В цикле while тоже может возникнуть коллизия переменных, если переменная с одним и тем же именем используется для каких-то целей как во внешнем цикле, так и во внутреннем. Вот пример.

```python
i = 0
while i < 3:
    print(f"Внешний: {i}")
    i = 0  # Сброс переменной из условия, если бы не коллизия, приводил бы к тому, что внешний цикл работает бесконечно
    while i < 2:
        print(f"  Внутренний: {i}")
        i += 1
    # Внутренний цикл завершился, i стало равно 2
    i += 1  # i станет 3, внешний цикл завершится.
```

#### break и continue
Еще у циклов есть две команды, о которых следует помнить: break и continue.

Команда break "ломает" цикл: как только программа на нее натыкается, она мгновенно выходит из цикла, не доделывая команды, которые идут за ним. Именно она позволяет закончить цикл, заданный через while True. Без этой команды такой цикл выполнялся бы бесконечно (или до тех пор, пока работа программы не будет прервана вручную), потребляя все больше и больше ресурсов, и мог бы даже заставить систему зависнуть или вылететь.

Команда continue выводит цикл из текущей итерации сразу на следующую: команды, которые за continue, не выполнятся, но цикл не остановится. Команда continue особенно удобна при работе со вложенными циклами.

Ниже даны примеры цикла while, выход из которого осуществляется только по команде break.

```python
while True:
    a = input()
    if a.isdigit():
        break
    print('Введите число!')
```

```python
while True:
    a = input()
    if a.isdigit():
        break
    if a.isspace():
        print('Нам не нужен пробел!')
        continue
    print('Вы ввели букву!')
```

Второй пример немного искусственный, однако на нем можно посмотреть, как работает команда continue.

И continue, и break могут использоваться как в цикле for, так и в цикле while, но чаще в последнем.

Нужно помнить, что если цикл снабжен блоком else, то при выходе из цикла по break этот блок не выполнится.

##Строки и методы работы с ними

Строка - это набор символов. В этом смысле строка неэлементарна и разложима; строчку можно перебрать по символам, а значит, она итерируемая: итерируемые объекты - такие, которые можно перебрать по частям. При этом строка неизменяемая! Мы можем обратиться к одному символу в строке, но перезаписать его не можем.

Существует несколько вариантов, как можно закавычивать строки:

    'Hello'
    "Hello"

    '''Hello'''
    """Hello"""

Теоретически любой вариант сообщает питону, что перед ним строка. Какие между этими способами различия?

- если у нас внутри строки встречается апостроф, удобнее использовать двойные кавычки;

- тройные кавычки позволяют строку разрывать переносами, а одинарные (в любом варианте) требуют, чтобы мы писали \n:


        'Hello\nworld'
        """Hello
        world"""

- а еще тройными кавычками часто обозначают комментарии в коде! Комментарии вообще мы уже видели: если комментарий - это просто приписка к строке кода, можно его отделить #, но это действует только до конца строки. Тройными кавычками выделяются длиннющие комментарии на много строк.

В связи с этим давайте познакомимся со специальными последовательностями в питоне. Некоторые символы в строках обозначают не сами себя, а что-то еще, например, перенос на новую строку обозначается условно как \n. Бэкслеш \ вообще служит для *экранирования*: то есть когда мы хотим, чтобы какой-то символ обозначал сам себя, мы перед ним можем поставить \, ну например:

```python
print('King\'s crown')  # мы хотим, чтобы апостроф в слове king's читался именно как апостроф, а не как конец строки
print('This is backslash: \\')  # бэкслэш экранирует сам себя: выведется только один \
```

Строка у нас состоит из символов, поэтому у нее есть длина! (Обратите внимание: числа, хотя они с точки зрения человека тоже могут состоять из нескольких цифр, составными для питона не являются. Поэтому для питона число - это *неитерируемый* цельный объект). Длину можно измерить с помощью функции len:

```python
len('qwerrty')
```

Старайтесь не называть свои переменные такими же именами, как функции питона, поэтому избегайте переменной с именем len :)

Символы в строке нумеруются, это называется индексация. Практически все (с о-очень редкими исключениями) вещи в питоне нумеруются с 0! Поэтому первый символ в строке идет под индексом 0, второй - 1 и так далее.

С понятием индекса связано понятие среза: срез - это часть строки (или списка).

Как с этими вещами работать:

```python
s = 'qweqweqweqwe'
print(s[0]) # выведет первый символ в строке
print(s[1:4]) # выведет все символы со 2 по 4
print(s[1:5:2]) # выведет все символы со 2 по 5 через один
print(s[:2]) # выведет все символы с начала до 2
print(s[1::2]) # выведет все символы со 2 до конца через один
```

Общий шаблон для среза такой:

    [start:stop:step]

Первое число - почти обязательное: если мы укажем только его, то возьмется один элемент с этим индексом.

Второе число может отсутствовать: это правая граница отрезка. В питоне все отрезки по умолчанию воспринимаются как [...), то есть, левая граница входит в отрезок, а правая нет.

Шаг - тоже необязательный.

Если мы пропускаем что-то из этого, то питон по умолчанию считает, что start = 0, stop = len(s), step = 1.

Индексация в строке устроена циклично! Индекс -1 означает 0 - 1 и ссылается на последний элемент в строке. Очень удобно, когда не знаем заранее длины строки! Шаг тоже вполне может быть отрицательным, тогда мы просто пойдем по строке задом наперед.

Какие арифметические операции можно выполнять со строками:

```python
print('hello' + ' world') # конкатенация
print('hello' * 3) # умножение на целое число
```

Строки можно сравнивать: строки сравниваются *посимвольно*. Это означает, что питон сперва сравнивает между собой первые символы в обеих строках, если они равны - то вторые и так пока не дойдет либо до конца строки, либо до различающихся символов.

Как питон сравнивает символы? Тут нужно немножко углубиться в понятие кодировки.

Помним, что для компьютера все данные - это на самом деле числа, и наши буквы тоже должны быть закодированы в числах. Весь вопрос в том, как их кодировать. Существуют разные варианты кодировок.

Сперва люди решили отводить на кодировку одной буквы один байт (8 бит). То есть, каждой букве в специальной табличке приписывался номер из восьми ноликов и единичек. Но сколько уникальных номеров можно так записать? У нас всего 2 возможных значения для бита (0 и 1), а в байте 8 бит, следовательно, уникальных комбинаций будет по правилам комбинаторики:

```python
2 ** 8
```

Так люди в первые 128 столбцов этой таблички положили стандартные символы типа пробела, цифры и латинский алфавит, а остальные 128 символов оставили для других алфавитов. Это было очень неудобно, потому что текст в кириллице, получается, занимал те же порядковые номера в таблице, что и текст на греческом языке, например.

Сама кодировка, к слову, называется [ASCII](https://ru.wikipedia.org/wiki/ASCII), можете на Википедии про нее почитать. В Windows, по крайней мере, до версии 10 включительно, все работало в кодировке cp1251: это аски с добавлением кириллицы в хвост. Если вы работаете локально на своем компьютере, проверить кодировку можно следующими командами:

```python
import locale

locale.getpreferredencoding()
```

В операционных системах Linux и macOS (и в колабе, потому что его сервера работают на Linux), однако, используется другая разновидность. Что такое UTF-8?

Ну, очевидно, 256 символов на все буквы всех алфавитов недостаточно, поэтому люди решили отвести целых 4 байта на символ. Если мы кодируем одну букву 4 байтами (то есть, 32 битами), то можно закодировать:

```python
2 ** 32
```

Очень много букв! До сих пор заняты еще не все позиции. Такая кодировка в общем случае называется Unicode, и табличку для нее можно посмотреть, например, [здесь](https://unicode-table.com/en/). В Юникод влезают все алфавиты на свете, я вас поздравляю. Там есть несколько разновидностей: utf-8, utf-16 и utf-32, но нам с вами в это можно не вникать, достаточно знать, что в кодировке utf-8 работает колаб, а также ноутбуки с системами MacOS и Linux.

Так вот, и питон тоже работает в utf-8, а значит, поддерживает любые кракозябры. Собственно, и сравнение символов происходит именно на основании этой таблицы: у какого символа порядковый номер по юникоду меньше, тот и меньше. Узнать, какой порядковый номер у символа, можно функцией ord():

```python
ord('b')
```

Теперь настало время поговорить про функции и методы.

Мы знаем, например, две самых необходимых функции ввода и вывода: print и input. Обратите внимание:

    print() # вызовет функцию

    s = print # воспримет только название функции и не вызовет ее, и после такого присваивания можно писать s(), потому что s - это то же, что print

То есть, если у нас есть функция, чтобы ее вызвать, нужно поставить круглые скобочки. В скобочках могут быть аргументы (они бывают обязательные и нет).

У print() нет обязательных аргументов, но она умеет их принимать. Какие можно ей передавать аргументы:

    print(*args, sep='', end='')
    (Это на самом деле не все, но пока обойдемся этими).

*args - позднее мы разберемся, что это означает, но это постоянно встречается в документации питона, поэтому пока запомните, что это означает "неопределенное количество аргументов, переданных позиционным способом", а "позиционный способ" означает, что мы просто их пихаем в определенном порядке (будем разбирать, когда будем учиться в функции).

Также мы знаем еще такие функции:

```python
int() # превращает все в целые числа (что может)
float() # превращает все в числа с плавающей точкой
bool() # превращает все в тип bool: True\False
str() # превращает все в строки
type() # выводит, к какому типу относится наша переменная
ord()
chr()
len()
```

Так вот, а бывают методы. Методы - это тоже функции, но в отличие от функций-функций, их можно применять только к конкретному типу данных. То есть, в print() мы можем передать все, что угодно, или вообще ничего не передать:

```python
print()
```

Он гордый и независимый. А вот метод обязательно должен применяться к какому-нибудь конкретному экземпляру. Например, у нас есть строка s, и мы хотим убрать у нее с краев лишние пробельные символы:

```python
s = '   hello   '
s.strip()
```

Видим: метод пишется через точку от объекта, к которому применяется (это может быть переменная или конкретное что-то, но сам по себе метод не существует). Точка в питоне обозначает принадлежность чего-то чему-то.

В остальном метод работает, как и любая функция, то есть:

- может принимать аргументы или не принимать
- может иметь именованные параметры (которые с именем пишутся, как sep и end в print)
- может что-нибудь возвращать, а может ничего не возвращать

В частности, методы строк всегда что-нибудь возвращают. Это потому, что строки у нас *неизменяемые* (как и все числа): мы не можем изменить строки по частям, только полностью перезаписать.

Про возвращения мы подробно еще не говорили, но идея такая: любая функция (и метод) может вернуть какой-нибудь объект после того, как отработает, а может вернуть пустоту. Например, print() свою работу делает, но ничего не возвращает, никаких новых объектов:

```python
s = print() # print ничего не возвращает! Он только выводит в печать что-то
type(s) # тип объекта s - None, то есть, ничего.
```

Когда мы попытались сохранить результат работы функции print() в переменную, в ней оказалась пустота.

А вот input() - и большинство других нам известных функций - возвращает какой-то объект. В частности, input() всегда возвращает строчку:

```python
s = input()
type(s)
```

Итак, какие методы мы с вами рассмотрели:

```python
# Возвращают строки
s.replace('a', 'b')  # заменяет все 'a' на 'b' (может несколько символов заменить сразу: 'ab' на 'c')
s.strip()  # отрезает все пробельные символы слева и справа. Если в скобочках указать какой-то символ, то будет отрезать его: s.strip('!') будет отрезать восклицательные знаки
s.rstrip()  # тоже отрезает, но только справа
s.lstrip() # отрезает только слева
s.upper()  # ВЕРХНИЙ РЕГИСТР
s.lower()  # нижний регистр
s.capitalize() # Первая заглавная буква
s.title()  # Все Буквы Заглавные
# Возвращают числа:
s.find('y')  # индекс первого нашедшегося слева такого символа. Если ничего не нашел, то -1
s.rfind('l') # индекс первого нашедшегося справа такого символа. Если ничего не нашел, то -1
s.index('y') # работает так же, как find, но если не нашел такой символ, то вывалит ошибку
s.count('l')  # посчитает, сколько раз в строке встретился символ (или символы)
# Проверки (возвращают bool)
s.startswith('qwerty')  # начинается ли строка на?.. эквивалентно s[:6] == 'qwerty'
s.endswith('uiop')  # заканчивается ли строка на?.. эквивалентно s[-4:] == 'uiop'
s.isupper() # строка состоит только из заглавных букв?
s.islower() # строка состоит только из строчных букв?
s.istitle() # в строке все слова начинаются на заглавную букву?
s.isdigit() # '1233'
s.isalpha() # 'asd'
s.isspace() # '  \xa0 \t\n
s.isalnum() # '123asd'
```

Для проверки, есть ли какой-то символ (или последовательность символов) в строке, можно еще использовать оператор in:

```python
's' in 'strike'
```

такое выражение вычисляется и возвращает True\False. Конечно, как и все остальные условные выражения, его разумно использовать только с переменными (с литералами мы и без питона можем сказать, что буква s в слове strike есть).

#### Структуры данных и коллекции в Python

Объекты в питоне, помимо типов, еще обладают некоторыми признаками, по которым их можно объединять в (пересекающиеся) группы. Таких признаков несколько:
- итерируемые
- Изменяемые / неизменяемые (hashable)
- вызываемые

Итерируемые объекты - такие, которые можно перебрать поэлементно. К ним относятся, например, строки.

Неизменяемые, они же хешируемые - такие, которые нельзя изменить по частям, а только перезаписать полностью. Это те, которые мы обсуждали все это время: числовые типа данных и строки. Хешируемыми они называются потому, что для того, чтобы отыскать такой объект в памяти, питон вычисляет т.н. hash-функцию и приписывает объекту определенный номер в таблице (хеш-таблице). Такой поиск осуществляется быстро, почти мгновенно, поэтому работа с неизменяемыми объектами экономит вычислительные ресурсы.

Вызываемые объекты - это функции. Их мы изучим позже.

Итак, мы переходим к *изменяемым* типам объектов. Основных их - три: *списки*, *множества* и *словари*. Сейчас мы будем заниматься списками и кортежами, а множества и словари оставим на потом.

Все объекты перечисленных типов являются *структурами данных*. Это означает, что каждый объект такого типа содержит в себе другие объекты, определенным образом организованные. Вся аналитика данных является работой с большими структурами данных, преобразованием одних структур в другие, оптимизацией поиска, хранения и передачи таких структур.

Также списки, кортежи, множества и словари называют *коллекциями*. Это название используется чаще, чем словосочетание "структуры данных", но оно менее точно, поскольку акцентирует только то, что объект содержит другие объекты, но не его внутреннюю организацию. Строка также является коллекцией, потому что состоит из отдельных символов. Существуют в питоне и другие типы коллекций, полный список вы можете посмотреть в документации библиотеки collections.

####Списки

Важный и постоянно используемый тип данных - это список (list). Список выглядит как набор каких-то элементов в квадратных скобках:

    [1, 2, 3] - это список int.
    ['1', '2', '3'] - это список str.
    [True, False, False] - это список bool.
    [[1], [2], [3]] - это список списков, каждый из которых содержит один int

Элементы могут быть не только одного, но и разных типов (смешанные списки).

Список - это итерируемый изменяемый объект. Все элементы в списке индексируются, точно так же, как в строке (следовательно, к спискам тоже применимы срезы).

Как можно создать новый список?

1. Прямо записать его (не забывайте, что объекты разных типов пишутся в питоне строго так, как это принято для конкретного типа: так питон понимает, что нам нужно число или строка):

        A = [1, 2, 3]

2. Завести пустой и добавлять в него элементы по одному с помощью метода append(). Пустой список создается либо просто квадратными скобочками, либо явно функцией list().

        A = []
        В = list()
        A.append(5) # A = [5]
        В.append('пять') # B = ['пять']

3. Преобразовать другой итерируемый объект:

        A = list('qwerty') #создает список из отдельных символов строки 'qwerty'
        B = list(input().split()) #создает список из слов введенной строки, разделяя их по пробелам

4. Заранее проинициализировать список дефолтными значениями (нулями или пустыми строками) и затем заполнять его. К такому списку можно обращаться по индексам, в отличие от пустого списка, где A[i] вызовет ошибку Index out of range.

```python
A = [0] * 5 #для списка длиной 5 элементов
A[2] = 1
A[4] = 2
print(A)
```

Какие операции можно выполнять со списками?

Как мы знаем, все арифметические и логические операторы применяются к числам; сложение, умножение и логические операторы также применяются к строкам (две строки можно конкатенировать в одну, а также одну строку умножить на int).

Списки тоже можно складывать и умножать на число. Эффект такой же, как у строк:

    [1, 2, 3] + [4, 5, 6] = [1, 2, 3, 4, 5, 6]
    [1, 2, 3] * 2 = [1, 2, 3, 1, 2, 3]

Обратите внимание, что, поскольку все элементы списка хранятся под уникальными номерами-индексами, в самом списке а) могут быть повторяющиеся элементы) б) порядок элементов строго определен.

Наконец к элементам списка можно применять все те штуки, которые применяются к типу данных, к которому относится элемент.

```python
A = [""] * 2
A[0] = 'qwerty'
A[0] = A[0].replace('q', 's')
A[1] = 'asdfgh'
print(A[0] + A[1])
```

**Методы списков**

1. Как вы уже видели, добавлять элементы в список можно с помощью метода list.append(elem): такой элемент всегда добавляется в конец списка. Заметьте, что, если в аргумент метода передать список, этот список станет элементом, а не добавится к исходному "на равных":

        A = [1, 2]
        B = [3, 4]
        A.append(B) = [1, 2, [3, 4]]

2. Есть метод, который работает так же, как +:

        A.extend(B)

3. Можно посчитать количество вхождений какого-то элемента в списке:

        A.count(elem)

4. Удалять элементы в списке можно по-разному.

        del A[i]
        A.pop(i) - не просто удалит i-ый элемент, но и вернет его (можно присвоить в переменную)
        A[i:i + 1] = [] - хитрый способ через срезы.

5. Для удаления по значению есть метод .remove(), он удаляет первое вхождение елемента в список. Если такого элемента в списке нет, получим ValueError.

        A.remove(elem)

6. Вставлять элементы тоже можно двумя способами:

        A.insert(index, elem)
        A[i:i] = [elem]

Будьте внимательны! Если вы используете итерацию по списку, то после удаления или вставки элемента индексы сдвигаются.


```python
A = [0, 1, 2, 3]
for elem in A: #сначала в списке 4 элемента, и мы перебираем их по порядку
    if elem <2: #условию удовлетворяют два первых элемента
        A.pop(elem) #удаляем по значению
print(A) #оказывается, что мы удалили только первый элемент, а второй пропустили из-за сдвига индексов
```

Еще есть два метода строк, связывающие их со списками: .split() и .join(). Метод .split() разрезает строку на несколько строк по указанному разделителю и помещает эти строки в список. По умолчанию разделителем считается пробел, но можно указать и другой, например, символ подчеркивания.

```python
string = "мама_мыла_раму"
A = string.split("_")
print(A)
```

Метод .join() превращает список в строку, объединяя элементы списка через указанный символ. Обратите внимание, что элементами списка могут быть не только строки, но и числа, которые питон превращает в строки перед объединением.

```python
A = ['4', 'черненьких', 'чумазеньких', 'чертенка']
print(" ".join(A))
```

К спискам также применяются следующие функции (не методы!):
- len(list) - возвращает длину списка (сколько в нем элементов)
- min(list) - возвращает самый маленький элемент
- max(list) - возвращает самый большой элемент
- sum(list) - возвращает сумму всех элементов, если все элементы - числа.

Как сравниваются между собой списки? Так же, как строки: поэлементно.

    [1, 2, 3] < [3, 4, 5]
    [1, 2, 3] < [1, 3, 4]
    [1, 1, 3] < [1, 1, 4]
    [1, 1, 1] == [1, 1, 1]

То есть, питон сперва сравнивает первые два элемента между собой, если они равны, то сравнивает следующие два элемента. Как только он доходит до момента, когда парные элементы не равны, то приписывает < или >.

**Копирование списков**

Напомню, что питон - язык, который берет на себя управление памятью за вас. С одной стороны, это очень здорово, потому что вам не приходится думать о том, как положить данные в оперативу; с другой стороны, питон считает себя умнее вас и поэтому специфично ведет себя с изменяемыми и неизменяемыми объектами, за что его не любят многие программисты С++ и других языков.

О чем это?

Когда у нас есть неизменяемый объект, например, число, и мы хотим записать его в другую переменную:

    x = 5
    y = x
    x += 1

y остается все равно 5, потому что в момент, когда мы сделали x = 6, переменная x просто стала указывать на новый объект (помните, что числа не изменяются, а просто теряются в памяти?)

Но когда у нас объект изменяемый, то операция присвоение B = A просто создаст еще одну ссылку на тот же самый объект, а не скопирует его.

```python
A = [1, 2, 3]
B = A
B[0] = 4
print(A)
```

То есть, объект "список" у нас один: но на него указывают две переменные. И если мы поменяем часть объекта по переменной В, то это отразится и в переменной А тоже!

Так сделано для того, чтобы глупенький программист не копировал списки (а они обычно огромные и занимают много памяти) зря. А нам, глупеньким программистам, приходится постоянно про это помнить, и случай в ячейке выше - еще самый простой!

Явно откопировать один список в другой можно при помощи метода copy:

    B = A.copy()

Также можно использовать для этой цели пустые срезы:

    B = A[:]

Но в общем и целом питон обычно прав, и не копируйте списки без особой нужды!

**Сортировка и переворачивание списков**

Списки можно сортировать двумя способами.

1. С помощью метода .sort()

        A = [2, 5, 3, 4]
        A.sort()

Метод изменяет список *на месте*: если вы напишете B = A.sort(), в переменную B положится None (пустота).

2. С помощью функции sorted()

        B = sorted(A)

Функция sorted() принимает на сортировку не только списки, но и любые итерируемые объекты, и возвращает их в виде списка. Поэтому результат работы функции можно и обычно нужно присваивать в переменную, а еще можно ей передавать строчки.

И у метода, и у функции есть один параметр: key, который умеет принимать функцию, по результатам которой будет сортировать.

Например, если мы хотим отсортировать список строчек не по алфавиту, а по длине строки:

```python
A = 'Richard of York gave battle in vain'.split()
print(A)
B = sorted(A, key=len)
print('B:', B)
A.sort(key=len, reverse=True)
print('A:', A)
```

В методе .sort() я здесь сразу использовала еще один параметр: reverse, который по умолчанию считается False, но если его явно указать как True, сами посмотрите, что будет.

Обратите внимание, что в параметре key функция передается без скобочек. Потому что здесь она не *вызывавется*, мы только сообщаем функции sorted(или методу sort), какую функцию нужно использовать. Функция вызывается уже где-то там в недрах сортировки. Конечно, мы должны использовать такие функции, которые умеют обрабатывать наш сортируемый тип данных и при этом что-нибудь возвращают! (Иначе зачем все это?..)

Аналогичным образом перевернуть список задом наперед можно двумя способами:

1. Методом reverse


    A.reverse()


2. Функцией reversed:


    B = reversed(A)

В случае с функцией нас, правда, ожидает маленькая засада, потому что функция возвращает не список, а какой-то другой, непонятный нам объект - *итератор*.

```python
A = [4, 6, 2, 7, 3, 9]
A.reverse()
print('New A:', A)
B = reversed(A)
print(B)
```

Однако этот объект можно легко превратить в список с помощью функции list().

Перевернуть список можно и с помощью среза, угадайте как.

#####Кортежи

Кортежам обычно в курсе программирования почти не уделяется внимания, потому что по функционалу они близки к спискам, единственное их отличие в том, что их нельзя изменять. Кортеж - это набор каких-то элементов; обычно они указываются в круглых скобочках, но можно задавать кортеж и без скобочек.

    T = (1, 2, 3)
    T = 1, 2, 3

Это создаст одинаковый кортеж в обоих случаях.

Из-за своей неизменности кортежи обрабатываются быстрее списков и удобны тогда, когда нам нужен некий набор констант.

У кортежей есть только два встроенных метода - это .count() для подсчета количества вхождений определенного элемента и .index() для поиска первого вхождения элемента. Они работают в точности так же, как одноименные методы списков.

В связи с кортежами следует упомянуть такую возможность питона, как *множественное присваивание*.

Внимательно считайте, сколько объектов стоит слева и справа!

    x, y = 4, 5 # присвоит 4 в x, а 5 в y
    x = 4, 5 # сделает в x кортеж (4, 5)
    x, y = 4 # вызовет ошибку!

Умный питон интерпретирует код справа от оператора присваивания по-разному в зависимости от того, сколько переменных мы ему указали слева.

Множественное присваивание можно использовать и со списками:

        x, y = [1, 2] # присвоит 1 в x, а 2 в y

С помощью множественного присваивания можно легко менять переменные местами, как показано ниже.

```python
x = 4
y = 5
x, y = y, x
print(x, y)
```

Питон сперва вычислит, что у него лежит справа (4, 5), а потом присвоит это в левую сторону. Так ничего не потеряется!

**Создание списков пар. Функции enumerate() и zip()**

Часто используемый в питоне вид кортежей - это упорядоченная пара объектов, в которой первый элемент пары принадлежит к какому-то одному типу данных, а второй - к другому. Например, в таком виде можно хранить пары из имени человека и номера его телефона. (Более продвинутый способ хранения таких пар предоставляют словари, с которыми мы познакомимся позже).

При этом из самих таких пар можно образовывать другие итерируемые объекты. Существует несколько удобных способов делать это. Мы сейчас рассмотрим два из них - функции enumerate() и zip().

Функция enumerate() предназначена для того, чтобы одновременно перебирать индексы и элементы какого-то итерируемого объекта. Ее удобно использовать с множественным присваиванием.

```python
fruits = ['banana', 'apple', 'plum', 'peach']
for i, elem in enumerate(fruits):
    print (f"Position {i+1} is {elem}.")
```

Функция zip() принимает в качестве аргументов два произвольных итерируемых объекта и соединяет и попарно их элементы. При этом, если длины объектов различаются, она выдает объект меньшей длины, и соединяет элементы с начала до того момента, пока в одном из наборов они не кончатся.

```python
A = [4, 6, 2, 7, 3, 9]
B = ("е","f", "g", "h")
print(list(zip(A, B)))
```

Обратите внимание, что zip() и enumerate() возвращают не списки, а опять-таки итераторы, поэтому если мы хотим увидеть результат их работы, нам нужно сначала преобразовать его, например, в список.

Если вам интересно, что такое итераторы и зачем они нужны, почитайте про них документацию. Но если в двух словах, итератор - это объект, который хранит не сами элементы, а способ их создания, потому что так экономится память и вычислительные ресурсы. Сами элементы вычисляются по мере необходимости, например, когда мы присваиваем их в какую-то переменную. Этот подход в программировании называется *стратегия ленивых вычислений*.
